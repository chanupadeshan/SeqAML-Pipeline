```python
from preprocessing.load import load_transactions
from preprocessing.features import add_features 
from preprocessing.leakage import (
    split_and_validate,
    get_split_summary
)


data_path = "../data/raw/HI-Medium_Trans.csv"

## load the data
df = load_transactions(data_path)

## add features
df = add_features(df)

## split the data into train and test sets
train_df, validation_df, test_df = split_and_validate(df, train_ratio=0.7, validation_ratio=0.15)
print("Temporal split completed successfully.")


summary = get_split_summary(train_df, validation_df, test_df)
summary.show()
```
